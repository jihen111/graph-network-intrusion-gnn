# Lancer le projet GNN sur Google Colab

1. **Exécution → Modifier le type d'exécution → GPU T4** (recommandé : environ 30 à 45 min ; sans GPU compte 1 h 30 à 2 h).
2. **Exécution → Tout exécuter.**
3. Dans la première cellule, clique sur **Choisir des fichiers** et sélectionne `graph-network-intrusion-gnn.zip`.
4. **Laisse l'onglet ouvert** jusqu'à la fin. Le navigateur télécharge alors `graph-network-intrusion-gnn-v2.zip` : c'est le projet complet avec les résultats. Envoie-le-moi.

Les données CIC-IDS2017 (~270 Mo) sont téléchargées automatiquement depuis Hugging Face.

```
from google.colab import files
import zipfile, os
uploaded = files.upload()  # choisis graph-network-intrusion-gnn.zip
name = next(iter(uploaded))
zipfile.ZipFile(name).extractall('/content')
os.chdir('/content/graph-network-intrusion-gnn')
print('Projet prêt :', os.getcwd())
```
